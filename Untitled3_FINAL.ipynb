## 23. Frequência de Termos nos Registros sem `MODO_FALHA`

```python
termos = [
    "partiu",
    "partida",
    "vibra",
    "ruído",
    "ruido",
    "trinca",
    "fissura",
    "corros",
    "deforma",
    "trip"
]

for termo in termos:
    quantidade = modos_ausentes["DESCRICAO_FALHA"].str.contains(
        termo,
        case=False,
        na=False
    ).sum()

    print(f"{termo}: {quantidade}")
```

## 24. Definição Inicial das Regras de Classificação

```python
def classificar_modo(descricao):
    texto = str(descricao).lower()

    if any(termo in texto for termo in ["vibração", "vibracao", "vibra", "rms"]):
        return "VIB"

    elif any(termo in texto for termo in ["ruído", "ruido", "barulho"]):
        return "NOI"

    elif any(termo in texto for termo in ["trinca", "fissura", "corros", "deforma", "desgaste estrutural"]):
        return "STD"

    elif any(termo in texto for termo in ["não partiu", "nao partiu", "não deu partida", "nao deu partida", "não iniciou", "nao iniciou", "não entrou em operação", "nao entrou em operacao"]):
        return "FTS"

    elif any(termo in texto for termo in ["perda de função", "perda de funcao", "parou", "trip", "falhou em operação", "falhou em operacao"]):
        return "FTF"

    else:
        return "AMBIGUO"
```

## 25. Teste das Regras de Classificação

```python
testes = [
    "Bomba PU-1031 não deu partida após acionamento.",
    "Vibração excessiva identificada no ventilador FAN-12006B.",
    "Ruído anormal no trocador de calor E-11010.",
    "Trinca na carcaça do motor M-2009B.",
    "Compressor K-3009B apresentou trips consecutivos durante o turno.",
    "Equipamento apresentou comportamento anormal durante operação."
]

for descricao in testes:
    print(descricao)
    print("Classificação:", classificar_modo(descricao))
    print()
```

## 26. Aplicação das Regras nos Registros sem `MODO_FALHA`

```python
modos_ausentes = modos_ausentes.copy()

modos_ausentes["MODO_SUGERIDO"] = modos_ausentes["DESCRICAO_FALHA"].apply(
    classificar_modo
)

modos_ausentes["MODO_SUGERIDO"].value_counts()
```

```python
def classificar_modo(descricao):
    texto = str(descricao).lower()

    # Vibração
    if any(termo in texto for termo in [
        "vibração", "vibracao", "vibra", "rms"
    ]):
        return "VIB"

    # Ruído
    elif any(termo in texto for termo in [
        "ruído", "ruido", "barulho"
    ]):
        return "NOI"

    # Deficiência estrutural
    elif any(termo in texto for termo in [
        "trinca", "fissura", "corros", "deforma",
        "desgaste estrutural", "degradação da fundação",
        "degradacao da fundacao", "estrutura comprometida",
        "fixação solta", "fixacao solta"
    ]):
        return "STD"

    # Falha ao partir
    elif any(termo in texto for termo in [
        "não partiu", "nao partiu", "n/ partiu",
        "não deu partida", "nao deu partida",
        "não respondeu à partida", "nao respondeu a partida",
        "n/ respondeu à partida",
        "não iniciou", "nao iniciou", "nao inicoui",
        "não energizou", "nao energizou",
        "não entrou em operação", "nao entrou em operacao",
        "não entrou em funcionamento", "nao entrou em funcionamento"
    ]):
        return "FTS"

    # Falha ao funcionar
    elif any(termo in texto for termo in [
        "perda de função", "perda de funcao",
        "perda de vazão", "perda de vazao",
        "perda de troca térmica", "perda de troca termica",
        "perda de capacidade",
        "parou", "desligou",
        "trip",
        "não atuou", "nao atuou",
        "não respondeu ao comando", "nao respondeu ao comando",
        "n/ respondeu ao comando",
        "não movimentou", "nao movimentou",
        "perdeu vedação", "perdeu vedacao",
        "ficou travada",
        "não recompôs", "nao recompos", "n/ recompôs",
        "operou com vazão abaixo", "operou com vazao abaixo"
    ]):
        return "FTF"

    else:
        return "AMBIGUO"
```

```python
modos_ausentes["MODO_SUGERIDO"] = modos_ausentes["DESCRICAO_FALHA"].apply(
    classificar_modo
)

modos_ausentes["MODO_SUGERIDO"].value_counts()
```

Casos em que o equipamento continua operando, porém com desempenho reduzido ou condição degradada, tendem a se enquadrar em `Degradação severa, equipamento funcional`.

Entretanto, descrições genéricas ou associadas apenas a sintomas, como vibração ou comportamento anormal, não permitem determinar o impacto com segurança sem análise adicional.

## 41. Definição Inicial das Regras de `IMPACTO_FALHA`

```python
def classificar_impacto(descricao):
    texto = str(descricao).lower()

    # Perda da função requerida
    if any(termo in texto for termo in [
        "não partiu", "nao partiu", "n/ partiu",
        "não deu partida", "nao deu partida",
        "não respondeu à partida", "nao respondeu a partida",
        "não iniciou", "nao iniciou", "n/ iniciou",
        "não atuou", "nao atuou",
        "não respondeu ao comando", "nao respondeu ao comando",
        "não movimentou", "nao movimentou",
        "parou", "desligou", "desarmou",
        "perda de função", "perda de funcao"
    ]):
        return "Falha: perda da função requerida"

    # Degradação severa, mas ainda funcional
    elif any(termo in texto for termo in [
        "vazão abaixo", "vazao abaixo",
        "perda de capacidade",
        "degradação", "degradacao",
        "desgaste estrutural",
        "corrosão", "corrosao",
        "trinca", "fissura",
        "deformação", "deformacao"
    ]):
        return "Degradação severa, equipamento funcional"

    # Casos em que há sintoma, mas sem evidência clara de perda funcional
    elif any(termo in texto for termo in [
        "vibração", "vibracao", "vibra",
        "ruído", "ruido", "barulho"
    ]):
        return "Equipamento funcional"

    else:
        return "AMBIGUO"
```

## 50. Estratégia Final de Classificação

A estratégia final combina regras semânticas de alta confiança com um modelo de texto utilizado apenas como fallback nos casos em que a descrição não permite uma decisão direta pelas regras.

Para `MODO_FALHA`, as regras semânticas têm prioridade quando identificam padrões claros. Nos casos restantes, é utilizado um classificador TF-IDF + Regressão Logística treinado apenas com exemplos em que a classificação original e a regra semântica concordam.

Para `IMPACTO_FALHA`, a correção automática é mais conservadora: somente evidências explícitas de perda funcional ou degradação severa sobrescrevem o valor original. Os valores ausentes que permanecem sem regra conclusiva são preenchidos pelo modelo de fallback.


```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
```

## 51. Regra de Alta Confiança para `IMPACTO_FALHA`


```python
def classificar_impacto_alta_confianca(descricao):
    texto = str(descricao).lower()

    termos_falha = [
        "não partiu", "nao partiu", "n/ partiu",
        "não deu partida", "nao deu partida",
        "não respondeu à partida", "nao respondeu a partida",
        "não iniciou", "nao iniciou", "n/ iniciou",
        "não entrou em operação", "nao entrou em operacao",
        "não atuou", "nao atuou", "n/ atuou",
        "não respondeu ao comando", "nao respondeu ao comando",
        "não movimentou", "nao movimentou",
        "não abriu", "nao abriu", "não fechou", "nao fechou",
        "ficou travada", "permaneceu bloqueada",
        "parou", "desligou", "desarmou", "desamrou",
        "trip", "trips", "perda de função", "perda de funcao"
    ]

    termos_degradacao = [
        "vazão abaixo", "vazao abaixo", "perda de capacidade",
        "degradação", "degradacao", "desgaste estrutural",
        "corrosão", "corrosao", "trinca", "fissura",
        "deformação", "deformacao", "vazamento", "histerese",
        "danos na vedação", "danos na vedacao"
    ]

    if any(termo in texto for termo in termos_falha):
        return "Falha: perda da função requerida"

    if any(termo in texto for termo in termos_degradacao):
        return "Degradação severa, equipamento funcional"

    return "AMBIGUO"
```

## 52. Treinamento dos Modelos de Fallback


```python
# Base de treinamento confiável para MODO_FALHA:
# somente registros em que o valor original concorda com a regra semântica.
treino_modo = df_trabalho[
    (df_trabalho["MODO_REGRA"] != "AMBIGUO") &
    (df_trabalho["MODO_FALHA"].notna()) &
    (df_trabalho["MODO_FALHA"] == df_trabalho["MODO_REGRA"])
].copy()

modelo_modo = Pipeline([
    ("tfidf", TfidfVectorizer(ngram_range=(1, 2), min_df=2, max_features=20000)),
    ("modelo", LogisticRegression(
        max_iter=1000,
        class_weight="balanced",
        random_state=42
    ))
])

modelo_modo.fit(
    treino_modo["DESCRICAO_FALHA"],
    treino_modo["MODO_FALHA"]
)

# Regra conservadora de impacto aplicada a toda a base.
df_trabalho["IMPACTO_ALTA_CONFIANCA"] = (
    df_trabalho["DESCRICAO_FALHA"]
    .apply(classificar_impacto_alta_confianca)
)

# Exemplos confiáveis de impacto:
# 1) original concorda com regra explícita;
# 2) equipamento funcional com modo VIB/NOI também confirmado pela regra de modo.
treino_impacto_explicito = df_trabalho[
    (df_trabalho["IMPACTO_ALTA_CONFIANCA"] != "AMBIGUO") &
    (df_trabalho["IMPACTO_FALHA"].notna()) &
    (df_trabalho["IMPACTO_FALHA"] == df_trabalho["IMPACTO_ALTA_CONFIANCA"])
].copy()

treino_impacto_funcional = df_trabalho[
    (df_trabalho["IMPACTO_FALHA"] == "Equipamento funcional") &
    (df_trabalho["MODO_REGRA"].isin(["VIB", "NOI"])) &
    (df_trabalho["MODO_FALHA"] == df_trabalho["MODO_REGRA"])
].copy()

treino_impacto = pd.concat(
    [treino_impacto_explicito, treino_impacto_funcional],
    ignore_index=True
).drop_duplicates(subset="ID")

modelo_impacto = Pipeline([
    ("tfidf", TfidfVectorizer(ngram_range=(1, 2), min_df=2, max_features=20000)),
    ("modelo", LogisticRegression(
        max_iter=1000,
        class_weight="balanced",
        random_state=42
    ))
])

modelo_impacto.fit(
    treino_impacto["DESCRICAO_FALHA"],
    treino_impacto["IMPACTO_FALHA"]
)

print("Registros confiáveis para treino de MODO_FALHA:", len(treino_modo))
print("Registros confiáveis para treino de IMPACTO_FALHA:", len(treino_impacto))
```

## 53. Consolidação Final de `MODO_FALHA` e `IMPACTO_FALHA`


```python
# MODO_FALHA final
df_trabalho["MODO_FINAL"] = df_trabalho["MODO_FALHA"]

# Quando a regra semântica é conclusiva, ela prevalece.
mask_modo_regra = df_trabalho["MODO_REGRA"] != "AMBIGUO"
df_trabalho.loc[mask_modo_regra, "MODO_FINAL"] = (
    df_trabalho.loc[mask_modo_regra, "MODO_REGRA"]
)

# Nos casos ainda sem classificação, utiliza o modelo de fallback.
mask_modo_fallback = df_trabalho["MODO_FINAL"].isna()
df_trabalho.loc[mask_modo_fallback, "MODO_FINAL"] = modelo_modo.predict(
    df_trabalho.loc[mask_modo_fallback, "DESCRICAO_FALHA"]
)

# IMPACTO_FALHA final
df_trabalho["IMPACTO_FINAL"] = df_trabalho["IMPACTO_FALHA"]

# Somente regras de alta confiança sobrescrevem valores existentes.
mask_impacto_regra = df_trabalho["IMPACTO_ALTA_CONFIANCA"] != "AMBIGUO"
df_trabalho.loc[mask_impacto_regra, "IMPACTO_FINAL"] = (
    df_trabalho.loc[mask_impacto_regra, "IMPACTO_ALTA_CONFIANCA"]
)

# Impactos ausentes sem regra conclusiva são preenchidos pelo fallback.
mask_impacto_fallback = df_trabalho["IMPACTO_FINAL"].isna()
df_trabalho.loc[mask_impacto_fallback, "IMPACTO_FINAL"] = modelo_impacto.predict(
    df_trabalho.loc[mask_impacto_fallback, "DESCRICAO_FALHA"]
)

print("MODO_FINAL")
print(df_trabalho["MODO_FINAL"].value_counts(dropna=False))

print("\nIMPACTO_FINAL")
print(df_trabalho["IMPACTO_FINAL"].value_counts(dropna=False))
```

## 54. Identificação do Tipo de Erro

Como a planilha final permite apenas um valor em `TIPO_DE_ERRO`, foi definida a seguinte prioridade quando um mesmo registro apresenta mais de um problema: `DUPLICADO` → `CAMPO_VAZIO` → `CLASSIFICACAO_INCORRETA` → `DESCRICAO_AMBIGUA` → `NENHUM`.


```python
# Duplicidade lógica do evento
mask_duplicado = df_trabalho.duplicated(
    subset=["DATA_FALHA", "SISTEMA", "TAG", "DESCRICAO_FALHA"],
    keep=False
)

# Campos originalmente vazios
mask_campo_vazio = (
    df_trabalho["MODO_FALHA"].isna() |
    df_trabalho["IMPACTO_FALHA"].isna()
)

# Classificação original divergente de regra considerada confiável
mask_modo_incorreto = (
    df_trabalho["MODO_FALHA"].notna() &
    (df_trabalho["MODO_REGRA"] != "AMBIGUO") &
    (df_trabalho["MODO_FALHA"] != df_trabalho["MODO_REGRA"])
)

mask_impacto_incorreto = (
    df_trabalho["IMPACTO_FALHA"].notna() &
    (df_trabalho["IMPACTO_ALTA_CONFIANCA"] != "AMBIGUO") &
    (df_trabalho["IMPACTO_FALHA"] != df_trabalho["IMPACTO_ALTA_CONFIANCA"])
)

mask_classificacao_incorreta = mask_modo_incorreto | mask_impacto_incorreto

# Descrição sem evidência semântica suficiente nas duas dimensões
mask_descricao_ambigua = (
    (df_trabalho["MODO_REGRA"] == "AMBIGUO") &
    (df_trabalho["IMPACTO_ALTA_CONFIANCA"] == "AMBIGUO")
)

df_trabalho["TIPO_DE_ERRO"] = np.select(
    [
        mask_duplicado,
        mask_campo_vazio,
        mask_classificacao_incorreta,
        mask_descricao_ambigua
    ],
    [
        "DUPLICADO",
        "CAMPO_VAZIO",
        "CLASSIFICACAO_INCORRETA",
        "DESCRICAO_AMBIGUA"
    ],
    default="NENHUM"
)

df_trabalho["REGISTRO_CORRETO_OU_INCORRETO"] = np.where(
    df_trabalho["TIPO_DE_ERRO"] == "NENHUM",
    "CORRETO",
    "INCORRETO"
)

df_trabalho["TIPO_DE_ERRO"].value_counts()
```

## 55. Construção da Planilha Final


```python
resposta_final = df_trabalho[["ID"]].copy()

resposta_final["MODO_FALHA"] = df_trabalho["MODO_FINAL"]
resposta_final["IMPACTO_FALHA"] = df_trabalho["IMPACTO_FINAL"]
resposta_final["REGISTRO_CORRETO_OU_INCORRETO"] = (
    df_trabalho["REGISTRO_CORRETO_OU_INCORRETO"]
)
resposta_final["TIPO_DE_ERRO"] = df_trabalho["TIPO_DE_ERRO"]

resposta_final.head()
```

A solução final utiliza uma abordagem híbrida e auditável: regras semânticas de alta confiança são utilizadas quando a descrição oferece evidência clara, enquanto modelos de classificação textual são empregados apenas como fallback para completar casos não resolvidos diretamente pelas regras.

A planilha final preserva os 10.000 IDs originais, utiliza exclusivamente os domínios definidos no processo seletivo e inclui a identificação de correção do registro e do principal tipo de erro encontrado.
